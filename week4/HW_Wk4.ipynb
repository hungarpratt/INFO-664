## Loops 
Instructions:

Make a list of numbers (ie [1,4,6,8,3]). Using a for loop, print an f-string for each number that states the square of each number in the list, by multiplying the number by its own value. IE: The square of {i} is {i*i}

```python
numbers=[3,4,7,8,10,11,14]

print(numbers)

for item in numbers:
    squared_number = item * item
    print(f"The square root of {item} is {squared_number}.")
```

## Conditionals
Instructions: 

Using the same list of numbers, nest an if statement in a for loop that goes through each number in the list. the if statement should check whether the number is even or odd, and will print {i} is even or {i} is odd respectively. Hint: use the % (modulo) operator to figure out if a number is even or odd. See this tutorial for more info.

```python
for item in numbers:
    if item % 2 == 0: #The modulo operator (%) returns the remainder of the item divided by 2. 
                      # If the result (the remainder) is 0, that means the item is an even number.
        print(f"{item} is an even number.")
    else:
        print(f"{item} is an odd number.")
```

## CSVs
Instructions: 
Using the MoMA CSV, count how many male artists are represented vs female artists vs non-binary artists. This will require you to
- Read the dataset as a dictionary list
- Filter that list to only include entries that have Gender data, by creating a new variable (ie artists_with_gender_data = [])
- Filter that list into three new lists using a for loop (which goes through each artist in the list), and an if/elif conditional that appends the given artist to the respective male_artists, female_artists, and non_binary_artists lists, based on whether the gender field matches. You may also want to create an other_artists field for entries that don’t match the criteria, using the else condition.
Produce a count for each of those lists using the len() function
Print the results into a cleanly-formatted f-string


```python
import csv

with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file) 
    data = list(reader) 
    
print(data[1]) # just testing to see what the headers are, and first row of data
```

```python
# I want to see all the different options for the gender field in the dataset, to know what to look for (and how things are spelled). 
# I used Claude to help me learn how to do this.

import pandas as pd

df = pd.read_csv("moma_artists.csv")

df["Gender"].unique()
```

```python
#Create a new variable of artists who have gender data (ie, exclude all artists with blanks in the gender field

artists_with_gender_data=[]

for artist in data: #look within the moma_artists dataset
    if artist["Gender"]!= "": # this pulls up all the artists who have some data in the Gender field (ie, who do NOT have blanks in that field)
        artists_with_gender_data.append(artist) # this creates a new list called artists_with_gender_data containing all of those artists with gender data

print(len(artists_with_gender_data)) #This provides a count of how many artists are contained in this new list
```

```python
female_artists = []
male_artists = []
nonbinary_artists = []
nonbinary_artists_expanded = [] # I want to make a list capturing artists with the 'non-binary' gender,
                                # as well as those with 'male (trans? ftm?)', 'gender non-conforming','transgender woman','female (transwoman)'

for artist in artists_with_gender_data: # So only looking for those artists who have gender data, per the artists_with_gender_data category
    if artist["Gender"] == "female":
        female_artists.append(artist)
    elif artist["Gender"] == "male":
        male_artists.append(artist)
    elif artist["Gender"] == "non-binary":
        nonbinary_artists.append(artist)
    else:
        nonbinary_artists_expanded.append(artist)
```

```python
print(f"There are {len(female_artists)} female artists.")
print(f"There are {len(male_artists)} male artists.")
print(f"There are {len(nonbinary_artists)} non-binary artists.")
print(f"There are {len(nonbinary_artists_expanded)} artists with different gender designations.")
```
