# Regular Expressions

```python
caption = "William Wegman, Cotto, 1970. Gelatin silver print, 7 7/8 × 7 3/4 in. (20 × 19.7 cm). Whitney Museum of American Art, New York; purchase with funds from the Mrs. Percy Uris Purchase Fund and the Photography Committee 92.14 © William Wegman"
```

Regular Expressions are a markup language used for identifying patterns in strings, in order to do complex searches. They use a bunch of specific special characters which represent different sequences to be extracted.

```python
import re #importing regular expression function/methods

match = re.search(r"[0-9]{4}", caption) # re.search tells Python to employ the RE function(?) and search for the items in the parentheses.
                                        # [0-9] means any digit 0,1,2,3,4,5,6,7,8,or 9. {4} means it has to be 4 digits. So [0-9]{4} means for digits
                                        # in a row, or a date (e.g. 1974).
                                        # caption means where to look for these results

print(match)
print(match.group(0))
```

```python
# '1970' is a string. To turn it into an integer (to do any math with it), we need to do the following:
year=int(match.group())
print(year)
print(year + 1)
print (year -10)
```

## Shorthand for using regular expressions and numbers 
Because digits are used so frequently in searches, regex has a shortcut way of dealing with them. 
- One way: type [0-0]{4} (which means looks for any digit - 0,1,2,3,4,5,...9 - that has four places
- Alternatively: type \d{4}

```python
match = re.search(r"\d{4}", caption) # r"\d{4}" is tje same aw writine r"[0-9]{4}
print(match)
print(match.group(0))
```

\d+ means find any digits. The "+" means one or more of the thing before me. 

\d+ finds any run of digits, of any length.

**re.findall()** is like re.search()

```python
print(re.findall(r"\d+", caption))
# That’s every number in the caption: the year, the inches, the centimetres (19.7 is split in 
# two around the decimal point), and the accession number 92.14. Asking for exactly four digits, 
# \d{4}, is what singles out the year.
```

## Finding the artist's name

match = re.search(r"^[^,]+", caption)
- The **^** at the begining of this means "start at the beginning of the string". It's telling Python to begin looking at the very beginning of the text block (in this case the caption)
- Anything in the [] means that Python should pull out **any** of the items in that block. For instance, if we saw {a,e,i,o,u] then Python could flag any time one of those vowels shows up. If we had **[,]** that would means look for the existence of a comma.
- If we have [^,] that means basically the opposite. It tells Python to look for any item **except** a comma. The ^ sign here inbetween the [] does something very different than the ^ earlier. [^,] tells Python to find anything that's NOT a comma, while ^ tells Python to start at the very beginning of the string.
- The **+** says find "one or more of the thing before me". So it's telling the computer to find all numbers and letters. So the pattern describes “everything from the start up to the first comma” so it's telling the computer to keep finding letters and numbers until it hits a comma.

```python
# The artist’s name is the first thing in the caption, followed by a comma. There’s no fixed shape to a name (how long is it, how many words), 
# but we do know where it ends: at the first comma. So we're telling the computer to find all the text (or numbers) from the beginnning of the string
# until the first comma.

match = re.search(r"^[^,]+", caption)
print(match)
print(match.group())
```

# Advanced: Finding the dimensions in centimetres (bonus)

The size appears twice, once in inches and once in centimetres: 7 7/8 × 7 3/4 in. (20 × 19.7 cm). We want the centimetre version, which is the one in parentheses, and we want the two numbers separately. The pattern for that is:

caption = "William Wegman, Cotto, 1970. Gelatin silver print, 7 7/8 × 7 3/4 in. **(20 × 19.7 cm)**. Whitney Museum of American Art, New York; purchase with funds from the Mrs. Percy Uris Purchase Fund and the Photography Committee 92.14 © William Wegman"

```python
match = re.search(r"\(([\d.]+) [x×] ([\d.]+) cm\)", caption)
```

```python
print(match.group())    # (20 × 19.7 cm)
print(match.group(1))   # 20
print(match.group(2))   # 19.7
print(match.groups())   # ('20', '19.7')

# HU Note: This is REALLY complicated! Craig's discussion in his Class Notes explains what's going on.
```

## Reading JSON files
We read JSON files the same way we read csvs, except they are naturally interpreted as lists of dictionaries

```python
import json #import JSON methods

with open("unesco_cultural_sites.json", "r", encoding="utf=8") as file:
    sites = json.load(file) # so going forward, let's call this dataset "sites"
    
print(type(sites))  # <class 'list'>
print(len(sites)) # 991 -- that's the number of items in this list (991 cultural sites)
print(sites[0].keys()) # For the first entry in the list, give me all the keys (basically the headers)
```

```python
print(sites[0]["name_en"]) # For the first item in the sites dataset, return the name (in English)
print(sites[0]["date_inscribed"])
```

```python
# Find the number of endangered sites
endangered_sites = []

for site in sites:
    if(site["danger"]== "True"):
        endangered_sites.append(site)

print(len(endangered_sites))
```

```python
for site in sites:
    if(site["danger"]== "True"):
        site_name = site["name_en"]
        site_state = site["region_code"]
        print(f"{site_name} is an endangered cultural heritage site in {site_state}.")
```

## APIs

APIs are used to query open data providers in real time by grabbing data from a server. The query is provided through a URL and the response is usually formatted as a JSON object. We will be looking at 2 different APIs today:
- The MET api, which can
    - Return object ids for search terms
    - Return details of a given object based on an object id
- NY TImes API, which can return lists of metadata based on search terms

```python
import requests #This brings in the functions needed to grab data from the web and deal with responses

# the variables make up the componenents of the uRL we will use to query the MET api

base_url = "https://collectionapi.metmuseum.org" #the base URL is always the same no matter what MET endpoint we're using
search_path = "/public/collection/v1.1/search" # the "endpoint" that deals with searching for objects in the collection
search_query = "?q=Antelope" # these are the parameters of the api query. q signifies search term and antelope is what we are searching for.
date_range = "&dateBegin=1900&dateEnd=2000"

url = f'{base_url}{search_path}{search_query}{date_range}'
search_response = requests.get(url)

print(url)   # https://collectionapi.metmuseum.org/public/collection/v1.1/search?q=Antelope
```

```python
object_path = "/public/collection/v1/objects/" # this is the path of the object's endpoint in the MET api

antelope_item = antelope_object_ids[0] # the first item in the list of objecct IDs above

url = f"{base_url}{object_path}{antelope_item}"

response_object = requests.get(url)

print(response_object.ok)
```
