### Adding new keys and values to a dictionary

My goal here is to look at the "running_time" value, extract a length in seconds by doing some simple math, and then creating a new key in each respective dictionary to represent the running time in seconds as an *int*.

```python
print(beatles_albums[0]["running_time"]) #accessing the running time key in the dictionary
```

```python
minute_string, second_string = beatles_albums[0]["running_time"].split(":")
print(minute_string)
print(second_string)
```

```python
total_seconds = int(minute_string) * 60 + int(second_string) # converting the minutes into seconds and adding seconds string to get a total number of seconds
print(total_seconds)
```

```python
beatles_albums[0]["running_time_in_seconds"] = total_seconds #adding a new key here. it automatically goes onto the end of the dictionary.
print(beatles_albums[0])
```

## Sorting dictionaries

When we are sorting lists we can use a simple sort method that puts items in alphabetical or numerical order. This is more difficult when dealing with dictionaries which have more diverse nested data. It's often the case that we don't merely want to alphabetize but to sort based on a specific nested quantity.

```python
print(central_park_birds)
```

```python
# sorted()function takes two arguments: one is the dataset to sort
# the other is the logic by which to sort
# in this case we use a lambda function to specify which key to prepare
# the optional third argument specifies whether the order should be ascending or descending
sorted_central_park_birds = sorted(central_park_birds, key=lambda item:item["count"], reverse = True) 

print(sorted_central_park_birds)
```
