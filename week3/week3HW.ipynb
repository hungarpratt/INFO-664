```python
print(mood_tracker)
```

## I would like to sort the order of this dictionary by mood, from lowest (1) to highest (5).

```python
sorted_moods = sorted(mood_tracker, key=lambda mood: mood["rating"])
```

```python
print(sorted_moods)
```

## Now I would like to reverse the order, from highest (5) to lowest (1).

```python
sorted_moods = sorted(mood_tracker, key=lambda mood: mood["rating"], reverse="True")
```

```python
print(sorted_moods)
```

## Now I would like to remove all the other fields so I am just looking at a list of mood ratings. (I used Claude to help me figure out how to do this..)

```python
ratings = [mood["rating"] for mood in sorted_moods]
print(ratings)
```

## Now I'd like to know the average of all my mood ratings over the course of the week.

```python
average = sum(ratings) / len(ratings)
print(average)
```

## I want to change the order from alphabetical (afternoon, evening, morning) to "time-based": morning, afternoon, evening. Again, I am consulting Claude to get some help with the code.

```python
order = {"morning": 1, "afternoon": 2, "evening": 3}
sorted_times = sorted(mood_tracker, key=lambda times: order[times["time"]])
ratings = [mood["time"] for mood in sorted_times]
print(ratings)
```
