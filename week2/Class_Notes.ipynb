## Data types in Python

Values in Python can be a number of different data types, such as **int, float, string, bool**

```python
print(type(5)) #INT (integer)
print(type(5.5)) # FLOAT (decimal point number)
print(type("5")) # STR (text data)
print(type(True)) # BOOL (boolean data -- true or false)
```

## Comparison Operators
Symbols in Python used to compare data, which evaluate to _True_ or _False_

```python
print(5>3) #This is a greater than operator
print(5==5) # This is an equal to operator
print (5!=5) # This is a not equal to operator
print (5<=3) # This is a less than or equal to operator
```

## Naming conventions and variables

- your variable name should represent what it is holding
- names CANNOT start with a number
- names CANNOT have a hyphen in them
- names CAN have underscores in them 
  - place_of_birth is a valid variable name using "snake case"
  - USE_UPPER_CASE to designate a constant variable that will not change after set (that's what we call a "house style" - it's a convention)
- names CANNOT be special words in the python language like "if" and "for"

## Converting data types

Data can be converted from one type to another using conversion functions

```python
pi = 3.14159
print(pi)

pi_as_string = str(pi) #str function converts to a string!
print(pi_as_string)
print(type(pi_as_string))

pi_as_float_again = float(pi_as_string)
print(pi_as_float_again)
print(type(pi_as_float_again))

print(pi_as_float_again > 3)
```

## Combining and concatenating strings

Strings can be combined in a number of ways. One is to use the *+* operator

```python
dog_name="Fido"
dog_breed="Labrador"
dog_age=4

#convert the int(dog_age) into a string before concatenating
my_dog_string="My dog's name is "+ dog_name + " and he is " + str(dog_age) 
print(my_dog_string)
```

We can also use [F-strings](http://www.google.com)

```python
my_dog_string = f"My dog's name is {dog_name}. He is {dog_age}, and he is a {dog_breed}."
print(my_dog_string)
```

## Converting string case

There are a bunch of built-in string methods for converting to upper, lower, capitalized and title case.

```python
print(dog_name.upper())
print(dog_name.lower())
print(my_dog_string.title())
print(my_dog_string.capitalize())
```

## Searching, counting, extracting

There are some string methods in Python that deal with searching for substrings, etc. One of those is the keywork **in**.

```python
print("cat" in my_dog_string) # True if the word "dog" is in the string"
print("dog" in my_dog_string) # True if the word "dog" is in the string"
```

```python
my_dog_string = my_dog_string + "I really love dogs and dogs are my best friends"
print(my_dog_string)
```

```python
dog_count = my_dog_string.count("dog")
print(f"The word 'dog' appears {dog_count} times.")
```

```python
cat_count = my_dog_string.count("cat")
print(cat_count)
```
