First I want to bring in the Moma artist dataset

```python
import csv # That will bring in all the CSV related functions

#The open function takes these 3 arguments:
#- file name or path
#- "r" for read
#- the encoding type (utf-8)

with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file) # read the file as dictionary data
    data = list(reader) # convert this object into a usable list called "data"
    print (data[0]) #print the first dictionary item
```

I want to make a list of all the names including "Mary"

```python
#step 1: make an empty list to hold artists with the name Mary

artists_named_mary = []

# step 2: create a for loop that goes through every item in the dataset

for artist in data:
    # step 3: extract first name
    first_name = artist['DisplayName'].split(" ")[0]
    # check if the first name matches Mary
    if "mary" in first_name.lower():
        # add our Mary to the new list
        artists_named_mary.append(artist)

print(len(artists_named_mary))

for mary in artists_named_mary[:10]:
    print(mary["DisplayName"])
```
