# Most Popular Birth Years for MOMA Artists
Assignment: MoMA's artists were born across hundreds of years. Which birth years show up the most often in the collection?
Write code that:
- Loads the whole dataset (you know how to do this already).
- **Builds a dictionary that counts how many artists share each BeginDate value — one key per year, one count per key.**
- Sorts that dictionary by count, largest first, and keeps just the top 10.
- Prints each of those years next to its count.

# (1) Load the Dataset

```python
import csv # That will bring in all the CSV related functions

with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file) # read the file as dictionary data
    data = list(reader) # convert this object into a usable list called "data"
    print (data[0]) #print the first dictionary item
```

# (2) Find Out How Many Artists Were Born in Each Year
This second step is going to be the hardest. The general approach should be as follows:

- (A) Create a dictionary called All_Years_Represented, which pulls out all the years where BeginDate does not equal 0.
- (B) **For each entry in All_Years_Represented, create lists of all the artists whose BeginDate is that year. (need a loop to do this).**
- (3) Sort the list by the most_popular_years and print out the first 10 (need to convert strings into integers if need be).

```python
# Find all the years in the dataset that have artists born then ("BeginDate")
# I guess it's not entirely necessary to exclude dates where there were no artists, since
# The list will ultimately be sorted by most popular to least popular birth years, but...

all_years_represented = []

for year in data: 
    if year['BeginDate'] != "0": # != means that BeginYear does NOT equal 0
        all_years_represented.append(year) # copy that year into my filtered list

for year in all_years_represented:
    print(year["BeginDate"])
```

```python
#Sort this list by the earliest BeginDate.
sorted(all_years_represented, key=lambda artist: artist["BeginDate"])
```
