## Step 2: Get the html once.

```python
import requests

url = "https://books.toscrape.com/catalogue/a-light-in-the-attic_1000/index.html" # This is a specific "product" page within the broader website.
headers = {"User-Agent": "INFO664-class-project (educational)"}

response = requests.get(url, headers=headers)
response.encoding = "utf-8" # I circled back to this line after I entered it to get rid of the weird characters in front of the price (thank you, Claude!)

with open("saved_page.html", "w", encoding="utf-8") as file:
    file.write(response.text)
```

This saves an .html file in my Jupyter Lab files.

## Step 3: Find the containers in my browser.

**Container:** `<article class="product_page">`
This seems to hold everything I care about: the image, price, description, and all the pieces of product information (product type, availability, etc.)

## Step 4: Walk through it with Beautiful Soup

```python
from bs4 import BeautifulSoup

with open("saved_page.html", "r", encoding="utf-8") as file:
    soup = BeautifulSoup(file.read(), "html.parser")

container = soup.find("article", class_="product_page")       # this is my specific container
children = container.find_all(True, recursive=False)

print(len(children))
```

```python
for i, child in enumerate(children):
    print(i, child.name, child.get("id"), "|", child.get_text(" ", strip=True)[:40])
    for j, grandchild in enumerate(child.find_all(True, recursive=False)):
        print("    ", j, grandchild.name, "|", grandchild.get_text(" ", strip=True)[:40])
```
